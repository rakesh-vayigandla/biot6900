## Part B — BioPython quickstart

```python
from Bio.Seq import Seq
dna = Seq("ATGGCCATTGTAATGGGCCGCTGA")
print("Length:", len(dna))
print("Protein:", dna.translate())
```

The given sequence is DNA. `.translate()` reads it in triplets (codons) and converts each codon into an amino acid, returning the final protein sequence. The `*` at the end signifies the stop codon.

## Part C — Biomedical Database Queries

```python
from Bio import Entrez
Entrez.email = "vayigandla.r@northeastern.edu"   
handle = Entrez.esearch(db="pubmed", term="AlphaFold protein structure", retmax=5)
record = Entrez.read(handle)
print("PMIDs:", record["IdList"])
```

I searched PubMed for "AlphaFold protein structure" and it returned 5 PMIDs (PubMed article IDs). Searching the literature matters because it shows what's already known about a protein — whether it's linked to the disease and whether others have tried to target it — before you invest in it as a drug target.

```python
import requests
r = requests.get("https://rest.uniprot.org/uniprotkb/P04637.json")
data = r.json()
name = data["proteinDescription"]["recommendedName"]["fullName"]["value"]
print("Protein:", name)
```

I queried accession P04637 and UniProt returned the protein "Cellular tumor antigen p53." UniProt matters because it gives you the protein's sequence, function, and disease links, the biological information you need to understand a target before working on it.

```python
import requests
r = requests.get("https://data.rcsb.org/rest/v1/core/entry/1TUP")
print("Title:", r.json()["struct"]["title"])
# the structure file itself:
# # https://files.rcsb.org/download/1TUP.pdb
```

I queried PDB ID 1TUP and it returned the structure titled "Tumor suppressor p53 complexed with DNA." Having a 3D structure matters for drug discovery because we can visualise the shape and the binding site of the molecule and design a drug so that it fits perfectly.

```python
import requests
base = "https://www.ebi.ac.uk/gwas/rest/api"
r = requests.get(f"{base}/singleNucleotidePolymorphisms/rs7412")
data = r.json()
print("Variant:", data.get("rsId"))
```

I queried variant rs7412 and the GWAS Catalog returned the variant ID rs7412. The GWAS Catalog links genetic variants with traits or diseases, which matters for finding a target because it gives you genetic evidence in humans that a gene is involved in a disease.
